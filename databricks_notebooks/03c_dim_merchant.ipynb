```python
df  = spark.table("silver_merchants")
display(df)
df.printSchema()
```

```python
keep_cols = [
    "merchant_id",
    "merchant_name",
    "category",
    "state",
    "city",
    "merchant_risk_level",
    "location",
    "merchant_rating",
    "merchant_status",
    "merchant_since"
]

keep_cols = [c for c in keep_cols if c in df.columns]
df = df.select(*keep_cols)
print(f" Kept {len(keep_cols)} columns")
```

```python
from pyspark.sql.functions import col
df = df.withColumnRenamed("merchant_id", "merchant_key")
print(" merchant_key set")
```

```python
from pyspark.sql.functions import current_timestamp
df = df.withColumn("_gold_updated_at", current_timestamp())
print(" Audit column added")
```

```python
df.write.format("delta").mode("overwrite").saveAsTable("dim_merchant")
print(f" dim_merchant: {spark.table('dim_merchant').count():,} rows")
```
