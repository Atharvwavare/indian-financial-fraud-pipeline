```python
df = spark.table("silver_transactions")
display(df)
```

```python
df.printSchema()
```

```python
fact_cols = [
    "transaction_id",
    "customer_id",
    "card_id",
    "merchant_id",
    "transaction_date",
    "transaction_time",
    "transaction_amount",
    "payment_method",
    "transaction_channel",
    "device_type",
    "transaction_status",
    "is_international",
    "fraud_flag",
    "fraud_reason",
    "merchant_risk_level",
    "merchant_category",
    "customer_state",
    "customer_city",
    "merchant_state",
    "merchant_city",
    "amount"
]

fact_cols = [c for c in fact_cols if c in df.columns]
df = df.select(*fact_cols)
print(f" Fact columns: {fact_cols}")
```

```python
from pyspark.sql.functions import col

rename_map = {
    "transaction_id": "transaction_key",
    "customer_id":    "customer_key",
    "card_id":        "card_key",
    "merchant_id":    "merchant_key",
}

for old, new in rename_map.items():
    if old in df.columns:
        df = df.withColumnRenamed(old, new)

print(" FKs renamed to match dims")
print(df.columns)
```

```python
from pyspark.sql.functions import date_format

df = df.withColumn(
    "date_key",
    date_format(col("transaction_date"), "yyyyMMdd").cast("int")
)

print(" date_key added")
df.select("transaction_date", "date_key").show(3)
```

```python
from pyspark.sql.functions import current_timestamp
df = df.withColumn("_gold_updated_at", current_timestamp())
print(" Audit column added")
```

```python
(df.write
   .format("delta")
   .mode("overwrite")
   .partitionBy("date_key")
   .saveAsTable("fact_transactions"))

print(f" fact_transactions: {spark.table('fact_transactions').count():,} rows")
```

```python
tables = [
    "dim_customer", "dim_card", "dim_merchant", "dim_date",
    "fact_transactions"
]
print(" Gold Layer — Star Schema")
print("=" * 45)
for t in tables:
    cnt = spark.table(t).count()
    print(f"   {t:<22} {cnt:>10,} rows")
```

```python
print("dim_merchant columns:")
print(spark.table("dim_merchant").columns)

print("\nfact_transactions columns:")
print(spark.table("fact_transactions").columns)
```

```python
spark.sql("""
    SELECT 
        m.merchant_name,
        f.merchant_category,
        SUM(f.transaction_amount) AS total_fraud_amount,
        COUNT(*) AS fraud_count
    FROM fact_transactions f
    JOIN dim_merchant m ON f.merchant_key = m.merchant_key
    WHERE f.fraud_flag = 1
    GROUP BY m.merchant_name, f.merchant_category
    ORDER BY total_fraud_amount DESC
    LIMIT 5
""").show(truncate=False)
```
