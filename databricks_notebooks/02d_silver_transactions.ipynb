```python
df = spark.table("bronze_transactions")
print(f"Rows: {df.count()}")
df.printSchema()
display(df)
```

```python
from pyspark.sql.functions import trim, col

for c, dtype in df.dtypes:
    if dtype == "string":
        df = df.withColumn(c, trim(col(c)))

print("Whitespace trimmed")
```

```python
rename_map = {
    "Transaction_ID":       "transaction_id",
    "Customer_ID":          "customer_id",
    "Card_ID":              "card_id",
    "Merchant_ID":          "merchant_id",
    "Transaction_Date":     "transaction_date",
    "Transaction_Time":     "transaction_time",
    "Transaction_Amount":   "transaction_amount",
    "Payment_Method":       "payment_method",
    "Transaction_Channel":  "transaction_channel",
    "Device_Type":          "device_type",
    "Transaction_Status":   "transaction_status",
    "Is_International":     "is_international",
    "Fraud_Flag":           "fraud_flag",
    "Fraud_Reason":         "fraud_reason",
    "Merchant_Risk_Level":  "merchant_risk_level",
    "Merchant_Category":    "merchant_category",
    "Customer_State":       "customer_state",
    "Customer_City":        "customer_city",
    "Merchant_State":       "merchant_state",
    "Merchant_City":        "merchant_city",
}

for old, new in rename_map.items():
    if old in df.columns:
        df = df.withColumnRenamed(old, new)

print("Columns after rename:")
print(df.columns)
```

```python
before = df.count()
df = df.dropDuplicates()
print(f" Dropped {before - df.count()} dupes")
```

```python
df = df.dropna(subset=["transaction_id", "customer_id"])
print(f"✅ Now {df.count()} rows")
```

```python
# Safety check — this must show lowercase columns
print("Columns after rename:")
for c in df.columns:
    print(f"  {c}")
```

```python
df = df.dropna(subset=["transaction_id", "customer_id"])
print(f" Now {df.count()} rows")
```

```python
before = df.count()
df = df.dropDuplicates()
print(f" Dropped {before - df.count()} dupes")
```

```python
from pyspark.sql.functions import to_date, col
from pyspark.sql.types import DecimalType, IntegerType

df = df.withColumn("transaction_amount", col("transaction_amount").cast(DecimalType(12, 2)))
df = df.withColumn("transaction_date", to_date(col("transaction_date"), "yyyy-MM-dd"))
df = df.withColumn("fraud_flag", col("fraud_flag").cast(IntegerType()))

print(" Types cast")
df.printSchema()
```

```python
before = df.count()
df = df.filter(col("transaction_amount") >= 0)
print(f" Dropped {before - df.count()} negatives")
```

```python
from pyspark.sql.functions import current_timestamp
df = df.withColumn("_silver_updated_at", current_timestamp())
print(" Audit column added")
```

```python
df.write.format("delta").mode("overwrite").saveAsTable("silver_transactions")
print(" Written")
```
