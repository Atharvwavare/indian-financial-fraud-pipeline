```python
df = spark.table("silver_customers")
df.printSchema()
```

```python
keep_cols = [
    "customer_id",
    "customer_name",
    "gender",
    "age",
    "marital_status",
    "occupation",
    "annual_income",
    "customer_segment",
    "state",
    "city",
    "account_type",
    "customer_since",
]

df = df.select(*keep_cols)
print(f"Kept {len(keep_cols)} columns: {keep_cols}")
```

```python
df = df.withColumnRenamed("customer_id", "customer_key")
print("customer_key set")
display(df.limit(4))
```

```python
from pyspark.sql.functions import current_timestamp
df = df.withColumn("_gold_updated_at", current_timestamp())
print(" Audit column added")
```

```python
df.write.format("delta").mode("overwrite").saveAsTable("dim_customer")
print(f" dim_customer: {spark.table('dim_customer').count():,} rows")
```
